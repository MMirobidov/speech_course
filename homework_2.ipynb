### Eval part (1 point)

Write balanced accuracy:
$$BAcc = \frac{1}{classes}\sum_{c = 1}^{classes} \frac{\sum_i^n I(y_i = p_i = c)}{\sum_i^n I(y_i = c)}$$

Where:
- $y_i$ -- target class for $i$ element
- $p_i$ -- predicted class for $i$ element

```python
# Get list of pairs (target_class, predicted_class)
from collections import defaultdict

def balanced_accuracy(items: list[tuple[int, int]]) -> float:
    total_per_class = defaultdict(int)
    correct_per_class = defaultdict(int)

    for target, pred in items:
        total_per_class[target] += 1
        if target == pred:
            correct_per_class[target] += 1

    result = 0
    num_classes = len(total_per_class)

    for cls in total_per_class:
        recall = correct_per_class[cls] / total_per_class[cls]
        result += recall

    result /= num_classes
    return result
        
```

```python
assert np.isclose(balanced_accuracy([(0, 0), (0, 0), (1, 1)]), 1.0)
assert np.isclose(balanced_accuracy([(0, 1), (1, 0)]), 0.0)
assert np.isclose(balanced_accuracy([(0, 0), (0, 0), (1, 0)]), 0.5)
assert np.isclose(balanced_accuracy([(0, 0), (1, 1), (0, 0), (0, 0), (1, 0), (0, 1)]), 0.625)
assert np.isclose(balanced_accuracy([(1, 1), (0, 1), (2, 2)]), 0.66666666666666)
```

```python
def get_num_params(model):
    result = 0
    for param in model.parameters():
        result += param.numel()
    return result


def stage(
    model: nn.Module,
    data: dataset.Dataset,
    opt: optim.Optimizer,
    batch_size: int = 256,
    train: bool = True
):
    loader = torch_data.DataLoader(
        data,
        batch_size=batch_size,
        shuffle=True,
        num_workers=LOADER_WORKERS,
        collate_fn=dataset.collate_fn
    )
    if train:
        model.train()
    else:
        model.eval()
    loss_sum, batches = 0.0, 0
    pred_pairs = []
    for X, Y in tqdm.tqdm(loader):
        if train:
            pred = model.forward(X.to(DEVICE))
            loss = F.cross_entropy(pred.squeeze(), Y.squeeze().to(DEVICE))
            opt.zero_grad()
            loss.backward()
            opt.step()
        else:
            with torch.no_grad():
                pred = model.forward(X.to(DEVICE))
                loss = F.cross_entropy(pred.squeeze(), Y.squeeze().to(DEVICE))
        loss_sum += loss.item()
        batches += 1
        with torch.no_grad():
            pred_pairs.extend(zip(
                Y.data.numpy().reshape(-1),
                torch.argmax(pred, dim=1).cpu().data.numpy().reshape(-1)
            ))
    return loss_sum / batches, balanced_accuracy(pred_pairs)


def train(
    model: nn.Module,
    opt,
    batch_size: int = 256,
    epochs: int = 10,
):
    train_losses, test_losses, train_accs, test_accs = [], [], [], []
    best_test_acc = 0.0
    best_test_loss = float('inf')
    for epoch in range(epochs):
        train_loss, train_acc = stage(model, trainset, opt, batch_size=batch_size)
        test_loss, test_acc = stage(model, testset, opt, batch_size=batch_size, train=False)
        train_losses.append(train_loss)
        train_accs.append(train_acc)
        test_losses.append(test_loss)
        test_accs.append(test_acc)
        if test_acc > best_test_acc:
            best_test_acc = test_acc
        if test_loss < best_test_loss:
            best_test_loss = test_loss
        clear_output()
        fig, axis = plt.subplots(1, 2, figsize=(15, 7))
        axis[0].plot(np.arange(1, epoch + 2), train_losses, label='train')
        axis[0].plot(np.arange(1, epoch + 2), test_losses, label='test')
        axis[1].plot(np.arange(1, epoch + 2), train_accs, label='train')
        axis[1].plot(np.arange(1, epoch + 2), test_accs, label='test')
        axis[0].set(xlabel='epoch', ylabel='CE Loss')
        axis[1].set(xlabel='epoch', ylabel='Accuracy')
        fig.legend()
        plt.show()
        print(f'Epoch {epoch + 1}.')
        print(f'Train loss {train_loss:.4f} | Train accuracy {train_acc:.4f}')
        print(f'Test  loss {test_loss:.4f} (best: {best_test_loss:.4f}) | Test accuracy {test_acc:.4f} (best: {best_test_acc:.4f})')
```

```python
class DepthwiseSeparable(nn.Module):
    def __init__(self, in_channels, out_channels, kernel_size, stride=1, padding=0):
        super().__init__()
        self.depthwise = nn.Conv2d(
            in_channels,
            in_channels,
            kernel_size=kernel_size,
            stride=stride,
            padding=padding,
            groups=in_channels
        )
        self.pointwise = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=1
        )

    def forward(self, X):
        return self.pointwise(self.depthwise(X))

class Model(nn.Module):
    def __init__(self, in_dim=FEATS, out_dim=N_CLASSES):
        super().__init__()

        self.sequential = nn.Sequential(
            DepthwiseSeparable(1, 64, kernel_size=3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.MaxPool2d(2),

            DepthwiseSeparable(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.MaxPool2d(2),

            DepthwiseSeparable(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(),
            nn.Dropout(0.2),

            nn.AdaptiveAvgPool2d((1,1))
        )

        self.fc = nn.Linear(256, out_dim)

    def forward(self, X):
        X = X.unsqueeze(1)  
        X = self.sequential(X)
        X = X.view(X.size(0), -1)
        X = self.fc(X)
        return X
```

```python
model = Model().to(DEVICE)
print("Number of parameters is ", get_num_params(model))

opt = optim.Adam(model.parameters())
```

```python
train(
    model,
    opt,
    batch_size=32,
    epochs=25
)
```
